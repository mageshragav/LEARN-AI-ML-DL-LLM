### Exercise: User Input and Calculator
Let's create a simple calculator that takes user input and performs basic arithmetic operations.

```python
# Simple Calculator

# Get user input
num1 = float(input("Enter first number: "))
num2 = float(input("Enter second number: "))
operation = input("Enter operation (+, -, *, /): ")

# Perform calculation
if operation == '+':
    result = num1 + num2
elif operation == '-':
    result = num1 - num2
elif operation == '*':
    result = num1 * num2
elif operation == '/':
    if num2 != 0:
        result = num1 / num2
    else:
        result = "Undefined (division by zero)"
else:
    result = "Invalid operation"

# Display result
print("Result:", result)
```
