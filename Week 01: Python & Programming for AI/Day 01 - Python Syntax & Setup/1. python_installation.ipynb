
#### 1. Python Installation
- **Windows**: You can download Python from the official website and run the installer. Make sure to check the option to add Python to your PATH.
- **Linux**: Python usually comes pre-installed on most Linux distributions. You can install the latest version using a package manager like `apt` for Ubuntu.

#### 2. Virtual Environments
- Virtual environments allow you to create isolated environments for Python projects. This helps in managing dependencies and avoiding conflicts between different project requirements.
- **Creating a Virtual Environment**:
  ```bash
  python -m venv myenv
  ```
- **Activating a Virtual Environment**:
  - **Windows**:
    ```bash
    myenv\Scripts\activate
    ```
  - **Linux**:
    ```bash
    source myenv/bin/activate
    ```

#### 3. Jupyter Notebook
- Jupyter Notebook is an open-source web application that allows you to create and share documents that contain live code, equations, visualizations, and narrative text.
- **Installation**:
  ```bash
  pip install jupyter
  ```
- **Running Jupyter Notebook**:
  ```bash
  jupyter notebook
  ```

#### 4. Visual Studio Code (VS Code)
- VS Code is a popular source-code editor made by Microsoft. It supports a wide range of programming languages and has a rich ecosystem of extensions.
- **Installation**: Download and install from the official website.
- **Extensions**: Install Python and Jupyter extensions for enhanced functionality.
