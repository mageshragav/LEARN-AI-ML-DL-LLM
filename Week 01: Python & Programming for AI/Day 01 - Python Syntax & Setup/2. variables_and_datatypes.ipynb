## Variables
Variables are like containers that store data. In Python, you don't need to<br> 
declare the type of a variable explicitly; it is inferred automatically. <br>
Here are some key points about variables:



```python
# Naming: Variable names can include letters, numbers, and underscores, 
# but they cannot start with a number. They are case-sensitive.
my_variable = 10
MyVariable = 20
_myvar = 30
```

```python
# Assignment: You assign a value to a variable using the equals sign (=).
x = 10
name = "Alice"

# Reassignment: You can change the value of a variable by assigning a new value to it.
x = 10
x = 20  # x now holds the value 20

# Multiple Assignment: You can assign values to multiple variables in a single line.
x, y, z = 10, 20, 30
```

### Data Types
Python has several built-in data types. Here are some of the most commonly used ones:


```python

# Numeric Types
# Integers (int): Whole numbers, positive or negative, without a decimal point.
x = 10

# Floating-Point Numbers (float): Real numbers with a decimal point.
y = 3.14

# Complex Numbers (complex): Numbers with a real and imaginary part.
z = 2 + 3j

# sequence Types
# Strings (str): A sequence of characters enclosed in single or double quotes.
name = "Alice"

# Lists (list): An ordered collection of items, which can be of different types.
my_list = [1, 2, 3, "apple", 4.5]

# Tuples (tuple): An ordered, immutable collection of items, which can be of different types.
my_tuple = (1, 2, 3, "banana", 5.6)

# Set (set): An unordered collection of unique items.
my_set = {1, 2, 3, "orange"}

# Mapping Types
# Dictionary (dict): A collection of key-value pairs, where keys are unique.
my_dict = {"name": "Alice", "age": 30, "city": "New York"}

# Boolean Type
# Boolean (bool): Represents one of two values: True or False.
is_active = True
```

### Type Conversion
You can convert between different data types using type conversion functions:

```python
# int(): Converts a value to an integer.
x = int(10.5)  # x is now 10

# float(): Converts a value to a floating-point number.
y = float(5)  # y is now 5.0

# str(): Converts a value to a string.
name = str(100)  # name is now "100"

# bool(): Converts a value to a boolean.
is_active = bool(1)  # is_active is now True

# list(): Converts a value to a list.
my_list = list((1, 2, 3))  # my_list is now [1, 2, 3]

# tuple(): Converts a value to a tuple.
my_tuple = tuple([1, 2, 3])  # my_tuple is now

# set(): Converts a value to a set.
my_set = set([1, 2, 3, 3, 4])  # my_set is now {1, 2, 3, 4}
```
